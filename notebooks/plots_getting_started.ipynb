# Plotting with fed3

## Import `fed3.plot`

The plotting package within fed3, `fed3.plot` is not imported by default.  Make an explicit call to import it:

```python
import fed3.plot as fplot
import panel as pn
from fed3.utilities import FED3DATA, agg_events, export_summary

pn.extension()
```

## Select data

Use the interactive tool below to select all csv files you want to analyze.

```python
fed_data = FED3DATA()
fed_data.select_data()
```

Use the interactive tool below to assign metadata to each csv data.
"group" and "animal" are pre-populated with dummy values.
If "start time" are not set, then no relative time subtraction will be performed.
Alternatively, you can use `fed_data.assign_metadata(rel_start=True)` to always enable the subtraction, where the time of first row in the data will be used as the start time if not explicitly set in the interactive tool.

```python
fed_data.assign_metadata()
```

Load the data after assigning all metadata.

```python
fed_data.load_data()
```

## Day night plot

Make day-night plot.
- Use `fplot.daynight_plot(..., bin_width="1hr")` to change time bin width.
- Use `fplot.daynight_plot(..., event="Event")` to change the event to aggregate.
- Use `fplot.daynight_plot(..., kind="bar")` to change to barplot.
- Use `fplot.fplot.daynight_plot(..., cummulative=True)` to enable cummulative aggregation.

```python
fplot.daynight_plot(
    fed_data.data_combined,
    bin_width="30min",
    event="Pellet",
    agg_grp="sum",
    anmcol="animal",
    kind="line",
    tcol="elps_time",
    cummulative=False,
)
```

## Plot Meal Size

Make meal size plot

```python
fplot.mealsize_hist(fed_data.data_combined)
```

## Export Summary

Export summary

```python
summary = export_summary(fed_data.data_combined)
summary.to_csv("summary.csv")
events = agg_events(fed_data.data_combined)
events.to_csv("events.csv")
```
